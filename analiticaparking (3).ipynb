```python
# esta función sirve para hacer un mejor recorrido de los datos
def transformar_datos(datos):
    datos_parking = df_parking_borrador.copy()

    # vamos a renombrar las columnas para facil llamado en la graficas
    datos_parking = datos_parking.rename(columns={
        "ID": "id",
        "Placas": "placa",
        "Tipo": "tipo",
        "Fecha ingreso": "fecha_ingreso",
        "Turno ingreso": "turno_ingreso",
        "Fecha salida": "fecha_salida",
        "Turno salida": "turno_salida",
        "Horas": "horas",
        "Minutos": "minutos",
        "Número de factura o comprobante": "factura",
    })


# convertimos los datos de las columnas de fecha en una fecha real y los de numeros en numeros reales, así como funciona en excel
    datos_parking["fecha_ingreso"] = pd.to_datetime(datos_parking["fecha_ingreso"], errors="coerce")
    datos_parking["fecha_salida"] = pd.to_datetime(datos_parking["fecha_salida"], errors="coerce")
    datos_parking["horas"] = pd.to_numeric(datos_parking["horas"])
    datos_parking["minutos"] = pd.to_numeric(datos_parking["minutos"])

# Convertir los datos de las columnas de fehcha en una fecha real y los de numeros en numeros reales
    datos_parking["fecha_salida"] = pd.to_datetime(datos_parking["fecha_salida"], errors="coerce")
    datos_parking["horas"] = pd.to_numeric(datos_parking["horas"])
    datos_parking["minutos"] = pd.to_numeric(datos_parking["minutos"])

# limpiar nuevamente las columnas para dejar unicamente la informacion util
    datos_parking = datos_parking[["id", "placa", "tipo", "fecha_ingreso", "turno_ingreso",
                   "fecha_salida", "turno_salida", "horas", "minutos", "factura"]]

# Ordenamos por fecha de ingreso y reorganizar el índice
    datos_parking = datos_parking.sort_values("fecha_ingreso").reset_index(drop=True)
    return datos_parking


df_parking7822 = transformar_datos(df_parking_borrador)
df_parking7822.info
print("Registros de vehículos:", len(df_parking7822))
df_parking7822.head()
```

```python
# Estamos definiendo una lista con los dias de la semana en español
DIAS = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
# funcion para devolver el nombre del día
def nombre_del_dia(fecha):
    """Recibe una fecha y devuelve el nombre del día de la semana."""
    return DIAS[fecha.weekday()]

# rangos de estadia de los clientes del parqueadero
def clasificar_duracion(minutos):
    """Clasifica una estadía según el tiempo que duró."""
    if minutos <= 60:
        return "Hasta 1 hora"
    elif minutos <= 180:
        return "De 1 a 3 horas"
    elif minutos <= 360:
        return "De 3 a 6 horas"
    elif minutos <= 720:
        return "De 6 a 12 horas"
    else:
        return "Más de 12 horas"

# conocer las jornadas de los clientes
def clasificar_jornada(hora):
    """AM si el vehículo entró antes del mediodía, PM si entró después."""
    if hora < 12:
        return "AM"
    else:
        return "PM"

# Descomponemos las fecha y hora en diferentes columnas para tener un mejor analisis
df_parking7822["fecha"] = df_parking7822["fecha_ingreso"].dt.date            # solo el día
df_parking7822["hora_ingreso"] = df_parking7822["fecha_ingreso"].dt.hour     # hora de 0 a 23
df_parking7822["hora_salida"] = df_parking7822["fecha_salida"].dt.hour

# identificar que dia de la semana es
df_parking7822["dia_semana"] = df_parking7822["fecha_ingreso"].apply(nombre_del_dia)

# convertimos las horas en minutos de la columna hora y le sumamos lo minutos de la solumna de los minutos,
# para tener todo en una misma unidad
df_parking7822["duracion_min"] = df_parking7822["horas"] * 60 + df_parking7822["minutos"]

# clasifica la duración en minutos para determinar la duración de los usuarios
df_parking7822["rango_duracion"] = df_parking7822["duracion_min"].apply(clasificar_duracion)

# calsificar en que jornada ingresan los usuarios
df_parking7822["jornada"] = df_parking7822["hora_ingreso"].apply(clasificar_jornada)

df_parking7822.info()
df_parking7822.head()
```

```python
# creamon una funcion para hacer el conteo de por tipo de vehiculo
def contar_valores(datos, columna):
    """Recorre una columna y cuenta cuántas veces aparece cada valor."""
    # crear un diccionario vacio llamado conteo, en el diccionario se va guardando los tipos de vehiculos y gracias al for se va sumando cuantas vaces aparece en el archivo
    conteo = {}
    for valor in datos[columna]:
        if valor in conteo:
            conteo[valor] = conteo[valor] + 1
        else:
            conteo[valor] = 1
    return conteo

# Usar la funcion ya creada con los valores de la columna tipo del archivo
conteo_tipos = contar_valores(df_parking7822, "tipo")

print("Vehículos por tipo:")
# el diccionario ya creado de los tipos de vehiculos que ingresan al parqueadero y la cantidad de veces
conteo_tipos.items()
# recorre el diccionario y le saca el porcentaje basandose en los datos
for tipo, cantidad in conteo_tipos.items():
    porcentaje = cantidad / len(df_parking7822) * 100
    print(" -", tipo, ":", cantidad, "(", round(porcentaje), "% )")
```

```python
def resumen_general(datos):
    """Recorre fila por fila y calcula indicadores del parqueadero."""
    total = 0
    suma_minutos = 0
    mas_larga = 0
    placa_mas_larga = ""
#
    for indice, fila in datos.iterrows():
        total = total + 1
        suma_minutos = suma_minutos + fila["duracion_min"]
        if fila["duracion_min"] > mas_larga:
            mas_larga = fila["duracion_min"]
            placa_mas_larga = fila["placa"]

    dias_con_movimiento = datos["fecha"].nunique()

    print("Periodo:", datos["fecha"].min(), "a", datos["fecha"].max())
    print("Total de ingresos:", total)
    print("Placas diferentes:", datos["placa"].nunique())
    print("Días con movimiento:", round(dias_con_movimiento))
    print("Promedio de vehículos por día:", round(total / dias_con_movimiento, 1))
    print("Duración promedio:", round(suma_minutos / total), "minutos")
    print("Duración mediana:", round(datos["duracion_min"].median()), "minutos")
    print("Estadía más larga:", round(mas_larga / 60, 1), "horas - placa", placa_mas_larga)


resumen_general(df_parking7822)
```

```python
def clientes_frecuentes(datos, minimo_visitas):
    """Devuelve las placas que han entrado al menos 'minimo_visitas' veces."""
    visitas = contar_valores(datos, "placa")
    frecuentes = {}
    for placa, cantidad in visitas.items():
        if cantidad >= minimo_visitas:
            frecuentes[placa] = cantidad
    return frecuentes


frecuentes = clientes_frecuentes(df_parking7822, 3)
print("Placas con 3 o más visitas:", len(frecuentes))

# Ordenamos de mayor a menor y mostramos las 10 primeras
top_10 = sorted(frecuentes.items(), key=lambda par: par[1], reverse=True)[:10]
for placa, cantidad in top_10:
    print(" -", placa, ":", cantidad, "visitas")
```

```python
#En esta grafica mostramos la cantidad de vehiculos y cuanto duran en el parqueadero
def grafico_duracion(datos):
  # Separamaos las estadías cortas (hasta 12 horas = 720 minutos) para poder graficarlas.
    # Las más largas harían el histograma muy ancho y casi no se vería la distribución.

    cortas = datos[datos["duracion_min"] <= 720]
    # Contamos las estadías largas restando: total de filas menos las cortas.
    largas = len(datos) - len(cortas)
    print("Estadías de más de 12 horas (no se muestran en el histograma):", largas)
    # Creamos la figura y definimos su tamaño (9 de ancho, 4.5 de alto).
    plt.figure(figsize=(9, 4.5))
    # Hacemos el histograma de la duración. Dividimos entre 60 para pasar de minutos a horas.
    # bins=24 con range=(0, 12) da barras de media hora (12 horas / 24 = 0.5 h cada una).
    # color y edgecolor son solo estéticos: barras azules con borde blanco para separarlas.
    plt.hist(cortas["duracion_min"] / 60, bins=24, range=(0, 12),
             color="steelblue", edgecolor="white")
    # Ponemos título y nombres a los ejes para que el gráfico se entienda solo.
    plt.title("Distribución del tiempo de permanencia (hasta 12 horas)")
    plt.xlabel("Horas en el parqueadero")
    plt.ylabel("Cantidad de vehículos")
    # Colocamos una marca en el eje X por cada hora entera (0 a 12).
    # Usamos range(0, 13) porque el 13 no se incluye.
    plt.xticks(range(0, 13))
    plt.show()


grafico_duracion(df_parking7822)
```

```python
# aquí generamos una tabla la cual nos permite conocer los datos de los tipos y su informacion general
def resumen_por_grupo(datos, columna):
    """Agrupa por una columna y calcula los indicadores de cada grupo."""
    resumen = datos.groupby(columna).agg(
      # redifinimos nuevas columnas para presentar la información
        ingresos=("id", "count"),
        placas_distintas=("placa", "nunique"),
        duracion_promedio=("duracion_min", "mean"),
        duracion_mediana=("duracion_min", "median"),
        duracion_maxima=("duracion_min", "max"),
    )
    resumen["porcentaje"] = resumen["ingresos"] / resumen["ingresos"].sum() * 100
    return resumen.round(1)


resumen_tipo = resumen_por_grupo(df_parking7822, "tipo")
resumen_tipo
```

8. Análisis para el dueño del parqueadero
8.1 Ocupación simultánea
¿Cuántos vehículos hay dentro del parqueadero al mismo tiempo?

Un vehículo está adentro en un momento dado si ya entró y todavía no ha salido. Recorremos el periodo hora por hora y contamos los vehículos que cumplen esa condición.

```python
def calcular_ocupacion(datos):
    """Recorre el periodo hora por hora y cuenta los vehículos que están adentro."""
    inicio = datos["fecha_ingreso"].min().floor("h")
    fin = datos["fecha_salida"].max().ceil("h")
    momentos = pd.date_range(inicio, fin, freq="h")

    lista_vehiculos = []
    for momento in momentos:
        ya_entro = datos["fecha_ingreso"] <= momento
        no_ha_salido = datos["fecha_salida"] > momento
        adentro = (ya_entro & no_ha_salido).sum()
        lista_vehiculos.append(adentro)

    ocupacion = pd.DataFrame({"momento": momentos, "vehiculos": lista_vehiculos})
    ocupacion["fecha"] = ocupacion["momento"].dt.date
    ocupacion["hora"] = ocupacion["momento"].dt.hour
    return ocupacion


ocupacion = calcular_ocupacion(df_parking7822)

fila_maxima = ocupacion.loc[ocupacion["vehiculos"].idxmax()]
print("Ocupación máxima:", fila_maxima["vehiculos"], "vehículos al mismo tiempo")
print("Ocurrió el:", fila_maxima["momento"])
ocupacion.head()
```

```python
def mapa_de_calor(datos):
    tabla = pd.crosstab(datos["dia_semana"], datos["hora_ingreso"]).reindex(DIAS, fill_value=0)

    plt.figure(figsize=(11, 4.5))
    plt.imshow(tabla.values, cmap="Blues", aspect="auto")
    plt.colorbar(label="Cantidad de ingresos")
    plt.xticks(range(len(tabla.columns)), tabla.columns)
    plt.yticks(range(len(tabla.index)), tabla.index)

    # Escribimos el número dentro de cada cuadro recorriendo filas y columnas
    limite = tabla.values.max() / 2
    for fila in range(len(tabla.index)):
        for columna in range(len(tabla.columns)):
            valor = tabla.values[fila, columna]
            if valor > limite:
                color_texto = "white"
            else:
                color_texto = "black"
            plt.text(columna, fila, valor, ha="center", va="center", color=color_texto)

    plt.title("Ingresos por día de la semana y hora")
    plt.xlabel("Hora de ingreso")
    plt.show()
    return tabla


tabla_calor = mapa_de_calor(df_parking7822)
```

```python
def momentos_mas_fuertes(tabla, cuantos):
    """Recorre la tabla del mapa de calor y devuelve los momentos con más ingresos."""
    lista = []
    for dia in tabla.index:
        for hora in tabla.columns:
            lista.append((tabla.loc[dia, hora], dia, hora))
    lista.sort(reverse=True)
    return lista[:cuantos]


print("Los 5 momentos con más ingresos:")
for cantidad, dia, hora in momentos_mas_fuertes(tabla_calor, 5):
    print(" -", dia, "a las", hora, "horas:", cantidad, "ingresos")
```

```python
def candidatos_mensualidad(datos, minimo_visitas):
    """Resume las visitas de cada placa y deja las que vienen seguido."""
    resumen = datos.groupby("placa").agg(
        tipo=("tipo", "first"),
        visitas=("id", "count"),
        dias_distintos=("fecha", "nunique"),
        horas_totales=("duracion_min", "sum"),
        primera_visita=("fecha", "min"),
        ultima_visita=("fecha", "max"),
    )
    resumen["horas_totales"] = (resumen["horas_totales"] / 60).round(1)
    resumen["horas_por_visita"] = (resumen["horas_totales"] / resumen["visitas"]).round(1)

    candidatos = resumen[resumen["visitas"] >= minimo_visitas]
    candidatos = candidatos.sort_values("visitas", ascending=False)
    return candidatos


candidatos = candidatos_mensualidad(df_parking7822, 3)
print("Placas con 3 o más visitas:", len(candidatos))
candidatos
```

```python
def resumen_recurrencia(datos):
    """Cuenta cuántas placas vinieron una sola vez y cuántas volvieron."""
    visitas = contar_valores(datos, "placa")
    una_vez = 0
    volvieron = 0
    for placa, cantidad in visitas.items():
        if cantidad == 1:
            una_vez = una_vez + 1
        else:
            volvieron = volvieron + 1

    total = una_vez + volvieron
    print("Placas que vinieron una sola vez:", una_vez, "(", round(una_vez / total * 100, 1), "% )")
    print("Placas que volvieron:", volvieron, "(", round(volvieron / total * 100, 1), "% )")


resumen_recurrencia(df_parking7822)
```

```python
def revisar_facturas(datos):
    """Busca números de factura repetidos y saltos en el consecutivo."""
    # El número 0 significa que la salida quedó SIN factura: se cuenta aparte
    sin_factura = 0
    numeros = []
    for numero in datos["factura"]:
        if numero == 0:
            sin_factura = sin_factura + 1
        else:
            numeros.append(numero)
    numeros.sort()

    repetidos = []
    faltantes = []
    for posicion in range(1, len(numeros)):
        anterior = numeros[posicion - 1]
        actual = numeros[posicion]
        if actual == anterior:
            repetidos.append(actual)
        elif actual > anterior + 1:
            # todos los números que hay entre el anterior y el actual
            for numero in range(anterior + 1, actual):
                faltantes.append(numero)

    print("Primera factura:", numeros[0], "- Última factura:", numeros[-1])
    print("Registros con factura:", len(numeros))
    print("Registros SIN factura (número 0):", sin_factura)
    print("Números repetidos:", len(repetidos))
    print("Números que faltan en el consecutivo:", len(faltantes))
    return repetidos, faltantes


repetidos, faltantes = revisar_facturas(df_parking7822)
print("Repetidos:", repetidos[:30])
print("Faltantes:", faltantes[:30])
```
